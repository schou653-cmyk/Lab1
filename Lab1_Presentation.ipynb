```python
from exceptions import (
    MeasurementUnitException,
    MissingValueException,
    TextFormatException,
)
from main import compute_BMI, parse_row

print("Setup successful!")
print("Setup successful!")
```

```python
# Test Case 1: Normal valid patient row
sample_valid = "1,10/2/2002,Wyn Grey,90,1.76"

attributes = parse_row(sample_valid)
bmi = compute_BMI(attributes[4], attributes[3])

print(f"Exam ID:        {attributes[0]}")
print(f"Weight:         {attributes[3]} kg")
print(f"Height:         {attributes[4]} m")
print(f"Calculated BMI: {bmi}")
```

```python
# Error Case 1: Height entered as 5.5 (feet instead of meters)
sample_unit_error = "522,2/27/2003,Xyra Thorn,90,5.5"

try:
    parse_row(sample_unit_error)
except MeasurementUnitException:
    print(f"Caught MeasurementUnitException successfully for row: {sample_unit_error}")
```

```python
# Error Case 2: Missing weight entry
sample_missing = "190,11/24/2002,Darin Vale,,1.58"

try:
    parse_row(sample_missing)
except MissingValueException:
    print(f"Caught MissingValueException successfully for row: {sample_missing}")
```

```python
# Error Case 3: Extra comma splits row into 6 values instead of 5
sample_format = '777,4/29/2003,"O\'Conor, Jared",78,1.53'

try:
    parse_row(sample_format)
except TextFormatException:
    print(f"Caught TextFormatException successfully for row: {sample_format}")
```
